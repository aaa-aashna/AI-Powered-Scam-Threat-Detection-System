```python
suspicious_keywords = [
    "urgent",
    "verify",
    "winner",
    "free",
    "click",
    "bank",
    "password",
    "claim",
    "limited",
    "offer",
    "prize",
    "account"
]
```

```python
def detect_suspicious_keywords(text):

    detected_keywords = []

    text = text.lower()

    for keyword in suspicious_keywords:

        if keyword in text:
            detected_keywords.append(keyword)

    return detected_keywords
```
